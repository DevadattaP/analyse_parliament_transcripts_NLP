```python
import langdetect
import sys
sys.path.append("../src")

from preprocessing.pdf_extractor import PDFExtractor
```

```python
def detect_language(text: str, sample_size=2000) -> str:
    """Detect document language with better sampling"""
    # Take samples from beginning, middle, and end
    text_length = len(text)
    if text_length > sample_size * 3:
        samples = [
            text[:sample_size],
            text[text_length//2 - sample_size//2:text_length//2 + sample_size//2],
            text[-sample_size:]
        ]
        sample = ' '.join(samples)
    else:
        sample = text
    
    try:
        return langdetect.detect(sample)
    except:
        return "unknown"
```

```python
def is_parliament_document(text, vocab_path="../data/vocabulary/parliament_vocab.txt", threshold=5):
    with open(vocab_path, "r", encoding="utf-8") as f:
        vocab = [line.strip().lower() for line in f.readlines()]
    
    text_lower = text.lower()
    for term in vocab:
        if term in text_lower:
            print(f"Found vocab term: {term}")
    match_count = sum(1 for term in vocab if term.lower() in text_lower)
    
    return match_count >= threshold, match_count, len(vocab)
```

```python
def validate_document(file_path):
    extractor = PDFExtractor()
    result = extractor.extract(file_path)
    text = "\n\n".join([seg["paragraph"] for seg in result["segments"]])
    language = detect_language(text)
    print(f"Detected language: {language}")
    is_valid, match_count, vocab_size = is_parliament_document(text)
    print(f"Is valid parliament document: {is_valid} (matched {match_count} out of {vocab_size} vocab terms)")
```
